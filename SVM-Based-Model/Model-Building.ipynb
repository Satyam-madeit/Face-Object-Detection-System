### CLEANING IMAGE DATA FOR TRAINING THE MODEL

```python
import os
import cv2
import numpy as np
from skimage.feature import hog
from sklearn.svm import LinearSVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import pickle
import random
```

```python
LFW = r'..\Datasets\LFW\lfw-deepfunneled\lfw-deepfunneled'
LFW_arr = []
for root, dirs, files in os.walk(LFW):
    for f in files:
        img = os.path.join(root, f)
        img_name = cv2.imread(img)
        resized_img = cv2.resize(img_name, (64, 64), interpolation=cv2.INTER_AREA)
        LFW_arr.append(resized_img)
LFW_arr = np.array(LFW_arr)
LFW_arr
```
