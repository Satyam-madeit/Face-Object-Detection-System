### CLEANING IMAGE DATA FOR TRAINING THE MODEL

```python
import os
import cv2
import numpy as np
from skimage.feature import hog
from sklearn.svm import LinearSVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import pickle
import random
```

```python
LFW = r'..\Datasets\LFW\lfw-deepfunneled\lfw-deepfunneled'
all_img_lfw = []
LFW_arr = []
for root, dirs, files in os.walk(LFW):
    for f in files:
        img = os.path.join(root, f)
        all_img_lfw.append(img)

random.shuffle(all_img_lfw)
for img in all_img_lfw:
    img_name = cv2.imread(img)
    resized_img = cv2.resize(img_name, (64, 64), interpolation=cv2.INTER_AREA)
    LFW_arr.append(resized_img)
    if len(LFW_arr) >= 6500: 
        break
LFW_arr = np.array(LFW_arr)
```

```python
def get_patch(img):
    height, width = img.shape[:2]
    x = random.randint(0, width - 64)
    y = random.randint(0, height - 64)
    patch = img[y:y + 64, x:x + 64]
    return patch


indoor = r'..\Datasets\Indoor Negative\indoorCVPR_09\Images'
all_img_indoor = []
indoor_arr = []
for root, dirs, files in os.walk(indoor):
    for f in files:
        img = os.path.join(root, f)
        all_img_indoor.append(img)

random.shuffle(all_img_indoor)
for img in all_img_indoor:
    img_name = cv2.imread(img)
    if img_name is None:
        continue
    heighti, widthi = img_name.shape[:2]
    if heighti >= 64 and widthi >= 64:
        patch = get_patch(img_name)
        indoor_arr.append(patch)
    if len(indoor_arr) >= 6500:
            break
indoor_arr = np.array(indoor_arr)
```
